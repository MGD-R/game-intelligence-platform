# Отчёт по обучению Entity Resolution

Notebook для итоговой презентации: очистка manual review, готовность к обучению, обучение модели и post-training metrics.

## Обязательные метрики для презентации

- Покрытие manual review: reviewed/skipped/pending counts.
- Баланс классов: positive/negative counts и positive rate.
- Качество данных: invalid display-name count и skipped invalid rows.
- Состав training set: manual labels vs weak labels vs synthetic negatives.
- Метрики baseline model: precision, recall, F1, ROC-AUC, PR-AUC.
- Manual-only validation metrics: те же метрики на human-reviewed test rows.
- Decision distribution: доли auto_merge/manual_review/no_merge после prediction.
