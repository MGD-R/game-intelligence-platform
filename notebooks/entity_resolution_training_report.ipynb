Notebook for the final presentation: manual review cleanup, pre-training readiness, model training, and post-training metrics.


## Required Presentation Metrics

- Manual review coverage: reviewed/skipped/pending counts.
- Class balance: positive/negative counts and positive rate.
- Data quality: invalid display-name count and skipped invalid rows.
- Training set composition: manual labels vs weak labels vs synthetic negatives.
- Baseline model metrics: precision, recall, F1, ROC-AUC, PR-AUC.
- Manual-only validation metrics: same metrics on human-reviewed test rows.
- Decision distribution: auto_merge/manual_review/no_merge rates after prediction.
